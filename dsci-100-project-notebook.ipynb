Visualization 1: Average Regulation Wins Percentage for Each Tourney Finish Level
If I want to figure out if regulation wins percentage can predict tourney finish, I should ensure that there is some relationship between these two variables first. Hence, a bar plot of the mean regulation wins percentage per tourney finish stage/level can be useful in determining/approximating the relationship. From the visualization, we can identify that there is a relationship between these two variables - teams that go onto later stages in the tournament seems to have higher mean regulation win percentage.

```r
#First creating a new tibble with the mean regulation wins percentage per tourney finish level
avg_reg_win_pct_per_tourney_finish<- tidy_womens_march_madness|>
group_by(tourney_finish)|>
summarize(mean_reg_wins_pct = mean(reg_wins_pct))     
avg_reg_win_pct_per_tourney_finish


#Creating the bar plot
avg_reg_win_pct_per_tourney_finish_bar<- avg_reg_win_pct_per_tourney_finish|>
ggplot(aes(x=fct_reorder(tourney_finish, mean_reg_wins_pct, .desc=TRUE), y= mean_reg_wins_pct))+    #reordered each level by descending, based on the mean regulation wins percentage
geom_bar(stat="identity")+                                                                          #set each bar value according to the y aesthetic (mean_reg_wins_pct)
labs(x="Tourney Finish Stage", y= "Average Regulation Win Percentage", title = "Average Regulation Win Percentage per Tourney Finish Stage")
options(repr.plot.width=10, repr.plot.height = 10)
avg_reg_win_pct_per_tourney_finish_bar
```

Visualization 2: Average Conference Wins Percentage for Each Tourney Finish Level
If I want to use conference wins percentage as a predictor for tourney finish, I should ensure that there is some relationship between them. Hence, a bar plot of the mean conference wins percentage per tourney finish level can be useful in determining the relationship. From the visualization, we can see for the most part, teams that go onto later stages in the tournament seem to have higher mean conference win percentage, with the primary exception of opening round loss teams having the highest mean conference win percentage.

```r
#First creating a new tibble with the mean conference wins percentage per tourney finish level
avg_conf_wins_pct_per_tourney_finish<- tidy_womens_march_madness|>
group_by(tourney_finish)|>
summarize(mean_conf_wins_pct = mean(conf_wins_pct, na.rm=TRUE)) #as seen in Task 4.2 there ARE NA values conference wins percentage
avg_conf_wins_pct_per_tourney_finish

#Creating a bar plot 
avg_conf_win_pct_per_tourney_finish_bar<- avg_conf_wins_pct_per_tourney_finish|>
ggplot(aes(x=fct_reorder(tourney_finish, mean_conf_wins_pct, .desc=TRUE), y= mean_conf_wins_pct))+    #reordered each level by descending, based on the mean regulation wins percentage
geom_bar(stat="identity")+                                                                          #set each bar value according to the y aesthetic (mean_conf_wins_pct)
labs(x="Tourney Finish Stage", y= "Average Conference Win Percentage", title = "Average Conference Win Percentage per Tourney Finish Stage")
options(repr.plot.width=10, repr.plot.height = 10)
avg_conf_win_pct_per_tourney_finish_bar
```
